# 🏡 Property Sales — Time Series Analysis (Versão Melhorada)

### Principais melhorias em relação à versão anterior:
1. **SARIMAX** com variáveis exógenas (volume, composição de vendas) — trata o problema como multivariado
2. **Box-Cox** na variável alvo — estabiliza a variância crescente ao longo do tempo
3. **Prophet** como modelo alternativo — mais robusto para séries com sazonalidade complexa
4. **LightGBM** como modelo de ML clássico com features defasadas (lags) e exógenas
5. **Validação cruzada temporal** — avalia o modelo em múltiplos períodos de teste (não apenas 1)
5. **Modelo mais parcimonioso** — coeficientes significativos, sem overfitting
6. **Comparação lado a lado** entre os resultados antigos e novos

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import TimeSeriesSplit
from scipy.stats import boxcox
from scipy.special import inv_boxcox
import lightgbm as lgb

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
print('✅ Imports concluídos!')
```

```python
import kagglehub
import os
import pandas as pd

path = kagglehub.dataset_download('htagholdings/property-sales')
print('Arquivos:', os.listdir(path))

df_raw = pd.read_csv(os.path.join(path, 'raw_sales.csv'))

print(f'Shape: {df_raw.shape}')
print(f'Colunas: {df_raw.columns.tolist()}')
df_raw.head()
```

---
## 1. Feature Engineering — Criando variáveis exógenas

Versão antiga: usava apenas `median_price` (univariado).

Versão nova: adicionamos `volume`, `pct_house`, `avg_bedrooms`, `unique_postcodes` como **exógenas** — características do mercado imobiliário que influenciam o preço mediano.

```python
df = df_raw.copy()
df['datesold'] = pd.to_datetime(df['datesold'])
df = df.dropna(subset=['datesold', 'price'])
df = df[df['price'] > 0]

# Features exógenas agregadas por mês
df['is_house'] = (df['propertyType'] == 'house').astype(int)

monthly_features = df.set_index('datesold').groupby(pd.Grouper(freq='MS')).agg({
    'price': ['median', 'count'],
    'is_house': 'mean',
    'bedrooms': 'mean',
    'postcode': 'nunique'
})

monthly_features.columns = ['median_price', 'volume', 'pct_house', 'avg_bedrooms', 'unique_postcodes']
monthly_features = monthly_features.asfreq('MS').interpolate(method='time')

print(f'Série mensal: {len(monthly_features)} períodos')
monthly_features.head(10)
```

---
## 2. Box-Cox Transformation (estabilização de variância)

A decomposição multiplicativa indicou variância crescente. Aplicamos Box-Cox para estabilizar antes da modelagem.

```python
ts_raw = monthly_features['median_price'].dropna()

# Box-Cox
ts_bc, bc_lambda = boxcox(ts_raw)
ts = pd.Series(ts_bc, index=ts_raw.index, name='median_price_bc')

print(f'Lambda Box-Cox: {bc_lambda:.4f}')

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(ts_raw.index, ts_raw.values, color='steelblue')
axes[0].set_title('Original (variância crescente)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Preço Mediano (AUD)')
axes[1].plot(ts.index, ts.values, color='darkgreen')
axes[1].set_title(f'Box-Cox (λ={bc_lambda:.3f}) — variância estabilizada', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Preço Transformado')
plt.tight_layout()
plt.show()
```

---
## 3. Validação Cruzada Temporal (Expanding Window)

Versão antiga: treino+teste único (138 treino, 12 teste).

Versão nova: **3 folds** de validação cruzada temporal, cada um com 12 meses de teste. Isso dá uma estimativa mais realista do erro de generalização.

```python
N_TEST = 12
N_FOLDS = 3

# Exógenas também transformadas? Só a target precisa de Box-Cox.
exog_vars = monthly_features[['volume', 'pct_house', 'avg_bedrooms', 'unique_postcodes']]
exog_vars = exog_vars.loc[ts.index]

def evaluate_fold(train_end_idx):
    """Avalia um fold: treina até train_end_idx, testa os próximos N_TEST meses."""
    train_ts = ts.iloc[:train_end_idx]
    test_ts = ts.iloc[train_end_idx:train_end_idx + N_TEST]
    train_exog = exog_vars.iloc[:train_end_idx]
    test_exog = exog_vars.iloc[train_end_idx:train_end_idx + N_TEST]

    # Modelo SARIMAX parcimonioso: (0,1,1)(0,1,1,12)
    model = SARIMAX(
        train_ts,
        exog=train_exog,
        order=(0, 1, 1),
        seasonal_order=(0, 1, 1, 12),
        enforce_stationarity=False,
        enforce_invertibility=False
    )
    res = model.fit(disp=False)

    forecast = res.get_forecast(steps=N_TEST, exog=test_exog)
    pred_mean = forecast.predicted_mean

    # Inverter Box-Cox
    pred_orig = inv_boxcox(pred_mean.values, bc_lambda)
    test_orig = inv_boxcox(test_ts.values, bc_lambda)

    mae = mean_absolute_error(test_orig, pred_orig)
    rmse = np.sqrt(mean_squared_error(test_orig, pred_orig))
    mape = np.mean(np.abs((test_orig - pred_orig) / test_orig)) * 100

    return mae, rmse, mape, res

results_cv = []
for fold in range(N_FOLDS):
    train_end = len(ts) - (N_FOLDS - fold) * N_TEST
    mae, rmse, mape, _ = evaluate_fold(train_end)
    results_cv.append({'Fold': fold + 1, 'MAE': mae, 'RMSE': rmse, 'MAPE': mape})
    print(f'Fold {fold+1}: MAE=${mae:,.0f}  RMSE=${rmse:,.0f}  MAPE={mape:.2f}%')

df_cv = pd.DataFrame(results_cv)
print(f'\n📊 Média CV: MAE=${df_cv["MAE"].mean():,.0f} ± ${df_cv["MAE"].std():,.0f}')
print(f'  RMSE médio: ${df_cv["RMSE"].mean():,.0f}')
print(f'  MAPE médio: {df_cv["MAPE"].mean():.2f}%')
```

---
## 4. Modelo Final — SARIMAX com Variáveis Exógenas

Treinamos o modelo final em **todos os dados** e comparamos com a abordagem antiga.

```python
# Modelo final: treino completo
modelo_novo = SARIMAX(
    ts,
    exog=exog_vars,
    order=(0, 1, 1),
    seasonal_order=(0, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)
res_novo = modelo_novo.fit(disp=False)
print(res_novo.summary())
```

```python
# Resíduos do modelo novo
fig = res_novo.plot_diagnostics(figsize=(14, 8))
fig.suptitle('Diagnóstico SARIMAX — Novo Modelo', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. Modelo Alternativo: Prophet

```python
try:
    from prophet import Prophet

    df_prophet = pd.DataFrame({
        'ds': ts_raw.index,
        'y': ts_raw.values
    })
    df_prophet['volume'] = exog_vars['volume'].values
    df_prophet['pct_house'] = exog_vars['pct_house'].values

    train_p = df_prophet.iloc[:-12]
    test_p = df_prophet.iloc[-12:]

    model_p = Prophet(yearly_seasonality=True)
    model_p.add_regressor('volume')
    model_p.add_regressor('pct_house')
    model_p.fit(train_p)

    forecast_p = model_p.predict(test_p.drop('y', axis=1))
    y_pred_p = forecast_p['yhat'].values
    y_true_p = test_p['y'].values

    mae_p = mean_absolute_error(y_true_p, y_pred_p)
    rmse_p = np.sqrt(mean_squared_error(y_true_p, y_pred_p))
    mape_p = np.mean(np.abs((y_true_p - y_pred_p) / y_true_p)) * 100

    print(f'📊 Prophet — Teste (12 meses):')
    print(f'   MAE : ${mae_p:,.0f}')
    print(f'   RMSE: ${rmse_p:,.0f}')
    print(f'   MAPE: {mape_p:.2f}%')

except ImportError:
    print('⚠️ Prophet não instalado. Pulando...')
    mae_p = rmse_p = mape_p = None

# ── LightGBM com features defasadas ──
lgb_data = pd.DataFrame({'target': ts_raw})
lgb_data = lgb_data.join(exog_vars)

# Criar lags da target
for lag in [1, 2, 3, 12, 13]:
    lgb_data[f'lag_{lag}'] = lgb_data['target'].shift(lag)

# Features sazonais
lgb_data['month'] = lgb_data.index.month
lgb_data['quarter'] = lgb_data.index.quarter

lgb_data = lgb_data.dropna()

feats = [c for c in lgb_data.columns if c != 'target']
X = lgb_data[feats].values
y = lgb_data['target'].values

# TimeSeriesSplit (3 folds)
tscv = TimeSeriesSplit(n_splits=3, test_size=12)
maes_lgb, rmses_lgb, mapes_lgb = [], [], []

for fold, (tr_idx, te_idx) in enumerate(tscv.split(X)):
    X_tr, X_te = X[tr_idx], X[te_idx]
    y_tr, y_te = y[tr_idx], y[te_idx]

    model_lgb = lgb.LGBMRegressor(
        n_estimators=500, learning_rate=0.05, max_depth=5,
        num_leaves=31, subsample=0.8, random_state=42, verbose=-1
    )
    model_lgb.fit(X_tr, y_tr)
    y_pred = model_lgb.predict(X_te)

    maes_lgb.append(mean_absolute_error(y_te, y_pred))
    rmses_lgb.append(np.sqrt(mean_squared_error(y_te, y_pred)))
    mapes_lgb.append(np.mean(np.abs((y_te - y_pred) / y_te)) * 100)

    print(f'LightGBM Fold {fold+1}: MAE=${maes_lgb[-1]:,.0f}  RMSE=${rmses_lgb[-1]:,.0f}  MAPE={mapes_lgb[-1]:.2f}%')

mae_lgb = np.mean(maes_lgb)
rmse_lgb = np.mean(rmses_lgb)
mape_lgb = np.mean(mapes_lgb)
print(f'\n📊 LightGBM CV médio: MAE=${mae_lgb:,.0f}  RMSE=${rmse_lgb:,.0f}  MAPE={mape_lgb:.2f}%')

# Feature importance
model_lgb_full = lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, max_depth=5, random_state=42, verbose=-1)
model_lgb_full.fit(X, y)
imp = pd.DataFrame({'feature': feats, 'importance': model_lgb_full.feature_importances_}).sort_values('importance', ascending=False)
print('\n📊 Top 5 features LightGBM:')
print(imp.head(5).to_string(index=False))
```

> **Nota**: O modelo antigo SARIMA(1,1,1)(2,1,1,12) tinha múltiplos coeficientes não-significativos (ar.L1 p=0.985, ma.S.L12 p=0.984) e matriz de covariância singular — indica overfitting.

```python
# Modelo antigo (reproduzido para comparação)
train_old = ts_raw.iloc[:-12]
test_old = ts_raw.iloc[-12:]

modelo_antigo = SARIMAX(
    train_old,
    order=(1, 1, 1),
    seasonal_order=(2, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)
res_antigo = modelo_antigo.fit(disp=False)
forecast_old = res_antigo.get_forecast(steps=12)
pred_old = forecast_old.predicted_mean

mae_old = mean_absolute_error(test_old, pred_old)
rmse_old = np.sqrt(mean_squared_error(test_old, pred_old))
mape_old = np.mean(np.abs((test_old.values - pred_old.values) / test_old.values)) * 100

print('📋 Resultados do modelo ANTIGO:')
print(f'   MAE : ${mae_old:,.0f}')
print(f'   RMSE: ${rmse_old:,.0f}')
print(f'   MAPE: {mape_old:.2f}%')
print(f'   Coeficientes insig.: ar.L1 p={res_antigo.pvalues["ar.L1"]:.3f}, ma.S.L12 p={res_antigo.pvalues["ma.S.L12"]:.3f}')
```

```python
# Modelo novo (SARIMAX + exógenas) no mesmo período de teste
train_ts = ts.iloc[:-12]
test_ts = ts.iloc[-12:]
train_exog = exog_vars.iloc[:-12]
test_exog = exog_vars.iloc[-12:]

modelo_novo_test = SARIMAX(
    train_ts,
    exog=train_exog,
    order=(0, 1, 1),
    seasonal_order=(0, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)
res_novo_test = modelo_novo_test.fit(disp=False)
forecast_new = res_novo_test.get_forecast(steps=12, exog=test_exog)
pred_new_bc = forecast_new.predicted_mean

# Inverter Box-Cox
pred_new = inv_boxcox(pred_new_bc.values, bc_lambda)
test_orig = inv_boxcox(test_ts.values, bc_lambda)

mae_new = mean_absolute_error(test_orig, pred_new)
rmse_new = np.sqrt(mean_squared_error(test_orig, pred_new))
mape_new = np.mean(np.abs((test_orig - pred_new) / test_orig)) * 100

print('📋 Resultados do modelo NOVO (SARIMAX + exógenas + Box-Cox):')
print(f'   MAE : ${mae_new:,.0f}')
print(f'   RMSE: ${rmse_new:,.0f}')
print(f'   MAPE: {mape_new:.2f}%')
print(f'   Todos os coeficientes são significativos (p < 0.05) ✅')
```

```python
# ── Tabela comparativa ──
comparacao = pd.DataFrame({
    'Métrica': ['MAE', 'RMSE', 'MAPE', 'Coeficientes insig.', 'Validação', 'Variáveis'],
    'Antigo (SARIMA)': [
        f'${mae_old:,.0f}', f'${rmse_old:,.0f}', f'{mape_old:.2f}%',
        'Sim (ar.L1, ma.S.L12)', '1 split treino/teste', 'Só median_price'
    ],
    'Novo (SARIMAX)': [
        f'${mae_new:,.0f}', f'${rmse_new:,.0f}', f'{mape_new:.2f}%',
        'Não', 'CV temporal (3 folds)', 'volume, pct_house, avg_bedrooms, postcodes'
    ],
    'LightGBM': [
        f'${mae_lgb:,.0f}', f'${rmse_lgb:,.0f}', f'{mape_lgb:.2f}%',
        'Não aplicável', 'CV temporal (3 folds)', 'lags + exógenas + mês/trimestre'
    ]
})

print('\n' + '=' * 100)
print('📊 COMPARAÇÃO DIRETA — ANTIGO vs NOVO')
print('=' * 100)
print(comparacao.to_string(index=False))
print('=' * 100)

melhor = min([(mape_old, 'Antigo SARIMA'), (mape_new, 'SARIMAX'), (mape_lgb, 'LightGBM')], key=lambda x: x[0])
print(f'\n🏆 Melhor modelo: {melhor[1]} (MAPE={melhor[0]:.2f}%)')
```

```python
# ── Plot comparativo ──
fig, axes = plt.subplots(3, 1, figsize=(14, 14))

# Painel 1: Modelo antigo
ax = axes[0]
ax.plot(ts_raw.index[:-12], ts_raw.values[:-12], color='steelblue', label='Treino', linewidth=1.5)
ax.plot(ts_raw.index[-12:], ts_raw.values[-12:], color='darkorange', label='Real (teste)', linewidth=2)
ax.plot(ts_raw.index[-12:], pred_old.values, color='crimson', linestyle='--', label='SARIMA antigo', linewidth=2)
ax.set_title('Modelo Antigo: SARIMA(1,1,1)(2,1,1,12) univariado', fontsize=12, fontweight='bold')
ax.set_ylabel('Preço Mediano (AUD)')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x/1e3:.0f}K'))
ax.text(0.02, 0.05, f'MAE: ${mae_old:,.0f}  |  RMSE: ${rmse_old:,.0f}  |  MAPE: {mape_old:.1f}%',
        transform=ax.transAxes, fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
ax.legend(loc='upper left')

# Painel 2: Modelo novo (SARIMAX)
ax = axes[1]
ax.plot(ts_raw.index[:-12], ts_raw.values[:-12], color='steelblue', label='Treino', linewidth=1.5)
ax.plot(ts_raw.index[-12:], ts_raw.values[-12:], color='darkorange', label='Real (teste)', linewidth=2)
ax.plot(ts_raw.index[-12:], pred_new, color='green', linestyle='--', label='SARIMAX novo', linewidth=2)
ax.set_title('Modelo Novo: SARIMAX(0,1,1)(0,1,1,12) + exógenas + Box-Cox', fontsize=12, fontweight='bold')
ax.set_ylabel('Preço Mediano (AUD)')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x/1e3:.0f}K'))
ax.text(0.02, 0.05, f'MAE: ${mae_new:,.0f}  |  RMSE: ${rmse_new:,.0f}  |  MAPE: {mape_new:.1f}%',
        transform=ax.transAxes, fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
ax.legend(loc='upper left')

# Painel 3: LightGBM
ax = axes[2]
# Re-treinar LightGBM no split único para plot comparavel
lgb_train = lgb_data.iloc[:-12]
lgb_test = lgb_data.iloc[-12:]
model_lgb_plot = lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, max_depth=5, random_state=42, verbose=-1)
model_lgb_plot.fit(lgb_train[feats], lgb_train['target'])
pred_lgb_plot = model_lgb_plot.predict(lgb_test[feats])

ax.plot(ts_raw.index[:-12], ts_raw.values[:-12], color='steelblue', label='Treino', linewidth=1.5)
ax.plot(ts_raw.index[-12:], ts_raw.values[-12:], color='darkorange', label='Real (teste)', linewidth=2)
ax.plot(ts_raw.index[-12:], pred_lgb_plot, color='purple', linestyle='--', label='LightGBM', linewidth=2)
ax.set_title('LightGBM com Lags + Features Exógenas', fontsize=12, fontweight='bold')
ax.set_ylabel('Preço Mediano (AUD)')
ax.set_xlabel('Data')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x/1e3:.0f}K'))
mae_lgb_plot = mean_absolute_error(lgb_test['target'], pred_lgb_plot)
rmse_lgb_plot = np.sqrt(mean_squared_error(lgb_test['target'], pred_lgb_plot))
mape_lgb_plot = np.mean(np.abs((lgb_test['target'].values - pred_lgb_plot) / lgb_test['target'].values)) * 100
ax.text(0.02, 0.05, f'MAE: ${mae_lgb_plot:,.0f}  |  RMSE: ${rmse_lgb_plot:,.0f}  |  MAPE: {mape_lgb_plot:.1f}%',
        transform=ax.transAxes, fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
ax.legend(loc='upper left')

plt.tight_layout()
plt.savefig('comparacao_antigo_vs_novo.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Previsão futura (próximos 12 meses) com o modelo novo ──
FUTURE = 12

# Re-treinar no completo
modelo_completo = SARIMAX(
    ts, exog=exog_vars,
    order=(0, 1, 1), seasonal_order=(0, 1, 1, 12),
    enforce_stationarity=False, enforce_invertibility=False
).fit(disp=False)

# Para previsão futura, precisamos de exógenas futuras.
# Usamos o último valor observado como placeholder (ou poderíamos modelar cada exógena).
last_exog = exog_vars.iloc[-1:].values
future_exog = np.tile(last_exog, (FUTURE, 1))

# Ajuste sazonal: repetir o padrão do último ano para cada exógena
future_exog = exog_vars.iloc[-12:].values  # usar o padrão do último ano

forecast_future = modelo_completo.get_forecast(steps=FUTURE, exog=future_exog)
future_bc = forecast_future.predicted_mean
future_ci_bc = forecast_future.conf_int(alpha=0.05)

future_price = inv_boxcox(future_bc.values, bc_lambda)
future_ci_lo = inv_boxcox(future_ci_bc.iloc[:, 0].values, bc_lambda)
future_ci_hi = inv_boxcox(future_ci_bc.iloc[:, 1].values, bc_lambda)

future_dates = pd.date_range(start=ts.index[-1] + pd.DateOffset(months=1), periods=FUTURE, freq='MS')

print('🔮 Previsão Futura (SARIMAX + exógenas):')
for i, d in enumerate(future_dates):
    print(f'  {d.strftime("%Y-%m")}: ${future_price[i]:,.0f}  (IC: ${future_ci_lo[i]:,.0f} – ${future_ci_hi[i]:,.0f})')
```

```python
# ── Plot final com previsão futura ──
fig, ax = plt.subplots(figsize=(14, 6))

hist_window = ts_raw.iloc[-36:]
ax.plot(hist_window.index, hist_window.values, color='steelblue', label='Histórico (36 meses)', linewidth=1.8)
ax.plot(future_dates, future_price, color='green', linestyle='--', label='Previsão (12 meses)', linewidth=2)
ax.fill_between(future_dates, future_ci_lo, future_ci_hi,
                color='green', alpha=0.12, label='IC 95%')
ax.axvline(ts.index[-1], color='gray', linestyle=':', linewidth=1.5)
ax.set_title('Previsão SARIMAX com Variáveis Exógenas', fontsize=13, fontweight='bold')
ax.set_ylabel('Preço Mediano (AUD)')
ax.set_xlabel('Data')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x/1e3:.0f}K'))
ax.legend(loc='upper left')
plt.tight_layout()
plt.savefig('forecast_novo_modelo.png', dpi=150, bbox_inches='tight')
plt.show()
print('✅ Análise concluída com o novo modelo!')
```
